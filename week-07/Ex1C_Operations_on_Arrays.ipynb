## Grocery Problem

```python
bag_1 = np.array([2,3,4,0])
bag_2 = np.array([4,0,1,2])
print(bag_1)
print(bag_2)
```

```python
added_b1 = np.sum(bag_1)
added_b2 = np.sum(bag_2)
print(f"The first bag contains {added_b1} items, and the second bag contains {added_b2} items.")
```

```python
#There are __ vegetables, __ fruits, __ boxes
# of snacks, and __ beverages in the two grocery bags
```

```python
total = bag_1 + bag_2
print(f"The are {total[0]} vegies, {total[1]} fruits, {total[2]} boxes of snacks, and {total[3]} beverages in the two grocery bags")
```

## Teamwork

```python
#Ali reported knocking on 5 doors on Monday, 4 on Tuesday, 6 on Wednesday, 4 on
# Thursday, and 2 on Friday
```

```python
ali = np.array([5,4,6,4,2])
print(ali)
```

```python
# Ben missed Monday because he was out of town. He reported knocking on 5 on
# Tuesday, 5 on Wednesday, 3 on Thursday, and 4 on Friday
```

```python
ben = np.array([0,5,5,3,4])
print(ben)
```

```python
# Carla reported knocking on 3 doors on Monday, 6 on Tuesday, 3 on Thursday, and 4
# on Friday. She was sick on Wednesday so missed that day
```

```python
carla = np.array([3,6,0,3,4])
print(carla)
```

```python
all_knocks = np.array([ali,ben,carla])
print(all_knocks)
```

```python
print(np.sum(all_knocks))
```

```python
daily_knock = np.sum(all_knocks,axis=0)
print(daily_knock)
```

```python
roll_knock = np.cumsum(daily_knock)
print(roll_knock)
```

```python
r_votes = roll_knock * 0.5
print(r_votes)
```
