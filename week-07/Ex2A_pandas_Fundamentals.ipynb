## Series

```python
# - Assign aa the list 1,2,3,4
# - Assign ab the tuple 5,6,7,8
# - Assign ac the set 9,10,11,12
# - Assign ad a dictionary with the following key-value pairs:'first': 13, 'second': 14, 'third': 15, 'fourth': 16
aa = [1,2,3,4]
ab = (5,6,7,8)
ac = {9,10,11,12}
ad = {'first':13,'second':14,'third':15,'fourth':16}
print(aa)
print(ab)
print(ac)
print(ad)
```

```python
# - ser1 should use aa
# - ser2 should use ab
# - ser3 should use ac
# - ser4 should use ad
ser1 = aa
ser2 = ab
ser3 = ac
ser4 = ad
print(ser1)
print(ser2)
print(ser3)
print(ser4)
```
