# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np 
```

## Looping Through Arrays

```python
arrA = np.array([[11,12,13],[14,15,16]])
print(arrA)
```

```python
for x in arrA:
    print(x*2)
```

```python
for x in arrA:
    for y in x:
        print(y*2)
```

### The first loop is going horizontal like on a x axis and the second loop is coming out vertical like on a y axis 

```python
loop_result = x * 2
for x in arrA:
    for y in x:
        print(x*2) # this just repeated the result 2 more times
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.concatenate((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
# stacked_array_2 = np.concatenate((joined_array_1, joined_array_2))
# print(stacked_array_2) # there was a error because they don't match
```

```python
another_joined_array = np.concatenate((joined_array_1,arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1,joined_array_1), axis=1)
print(another_2)
```

```python
h = np.hstack((arrA, arrA))
v = np.vstack((arrA, arrA))
d = np.dstack((arrA, arrA))

print(h)
print(v)
print(d)
```

## Splitting Arrays

```python
splitsky = np.split(joined_array_1, 2)
print(splitsky) # theres 2 arrays
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, "odd")
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7,6],[13,5],[7,3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
```

```python
z = arrB < 7
```

```python
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7 
```

```python
arrD = arrB[y]
print(arrD)
```
