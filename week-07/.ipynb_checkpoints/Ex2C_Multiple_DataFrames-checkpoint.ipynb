## Basic Aggregations

```python
data = {'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East',
'West', 'North'],
'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]}
```

```python
df_sales = pd.DataFrame(data)
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
```

```python
df_sales
```

```python
total_units = df_sales['Units_Sold'].sum()
total_revenue = df_sales['Revenue'].sum()
```

```python
print("Total Units Sold:", total_units)
print("Total Revenue:", total_revenue)
```

```python
avg_units = df_sales['Units_Sold'].mean()
avg_revenue = df_sales['Revenue'].mean()
```

```python
print("Average Units Sold:", avg_units)
print("Average Revenue:",avg_revenue)
```

## Grouping Data

```python
product_group = df_sales.groupby('Product')[['Units_Sold','Revenue']].sum()
product_group
```

```python
region_group = df_sales.groupby('Region')[['Units_Sold','Revenue']].mean()
region_group
```

```python
product_region_group = df_sales.groupby(['Product','Region'])[['Units_Sold','Revenue']].sum()
product_region_group
```

## Merging DataFrames

```python
product_data = {'Product': ['A', 'B', 'C'],
'Category': ['Electronics', 'Furniture', 'Appliances']}
```

```python
df_products = pd.DataFrame(product_data)
```

```python
df_merged = pd.merge(df_sales, df_products, on='Product')
df_merged
```

```python
category_revenue = df_merged.groupby('Category')['Revenue'].sum()
category_revenue
```
