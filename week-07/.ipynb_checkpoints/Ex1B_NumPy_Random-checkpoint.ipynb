# Exercise 1.B NumPy Random

```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1,21, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1,21, size=(1,2))
print(r_2)
```

```python
r_3 = np.random.rand(2,1)
print(r_3)
```

```python
r_4 = np.random.randint(20,101, size=(4,5))
print(r_4)
```

```python
r_5 = np.random.rand(4,5)
print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array(["Julissa","Longley","Angel","David","Tracy","Shay","El","Onur","Fuat","Maritza","Jimmy","Paul"])
print(students)
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size =(3,4))
print(breakout_groups) #I don't think this would be effective because it repeats names
```

```python
p_val = [0,0.03,0.04,0.05,0.06,0.07,0.07,0.08,0.10,0.10,0.10,0.30]
call_on_weighted = np.random.choice(students, size=5, p=p_val)
print(call_on_weighted) # I did the p_vals like that because everytime I got a error that they need to add to 1.00
```

 ## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4,5))
print(bell_curve)
```
