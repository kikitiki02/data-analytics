# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.title("Ex graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

## Styling Plots

```python
plt.plot(x,y, linestyle='--',color='red')
plt.show()
```

```python
plt.plot(x,y, linestyle='--',color='red',marker='s',markersize=10,markerfacecolor='green',markeredgecolor='green')
plt.show()
```

```python
plt.plot(x,y, linestyle='--',color='pink',marker='$♥$',markersize=10,markerfacecolor='black',markeredgecolor='pink')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
```

```python
plt.figure(figsize=(6,4))
plt.bar(products,sales,color='skyblue',width=0.5)
plt.title('Sales of Different Products')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
x = np.random.rand(50)
```

```python
y = np.random.rand(50)
```

```python
sizes = np.random.rand(50) * 100
```

```python
colors = np.random.rand(50)
```

```python
plt.figure(figsize=(6,4))
plt.scatter(x,y,s=sizes,c=colors,cmap='viridis')
plt.title('Scatter Plot of Random Data')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.colorbar(label='Color Scale')
plt.show()
```

## Subplots

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

```python
random_data = np.random.randn(1000)
```

```python
fig, (ax1,ax2) = plt.subplots(1,2,figsize=(12,5))
ax1.plot(x,y)
ax1.set_title('Sine Wave')
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)')

ax2.hist(random_data, bins=30)
ax2.set_title('Histogram of Random Numbers')
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')
plt.show()
```
